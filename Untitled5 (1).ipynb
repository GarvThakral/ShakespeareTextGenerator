```python
ds = load_dataset("Trelis/tiny-shakespeare")
maxInDs = 0
for x in ds['train']:
  sentence_array = re.findall(r"\b\w+(?:'\w+)?\b|[.!?,-]", x['Text'].lower())
  maxInDs = max(len(sentence_array),maxInDs)
maxInDs
```

```python
newVocab = []

def create_new_vocab(sentence,newVocab):
    sentence_array = re.findall(r"\b\w+(?:'\w+)?\b|[.!?,-]", sentence.lower())
    new_words = [x for x in sentence_array if x not in newVocab]
    newVocab += list(set(new_words))  # add uniques

# build vocab
ds['train'].map(lambda x : create_new_vocab(x["Text"], newVocab))
ds['test'].map(lambda x : create_new_vocab(x["Text"], newVocab))

# insert special tokens at fixed IDs
newVocab.insert(0, "<EOS>")   # id=0 or 3? we’ll fix ordering below
newVocab.insert(0, "<BOS>")
newVocab.insert(0, "<UNK>")
newVocab.insert(0, "<PAD>")

# final ordering:
# <PAD>=0, <UNK>=1, <BOS>=2, <EOS>=3
```

```python
vocab_size = len(newVocab)
index_to_word = {i:k for i,k in enumerate(newVocab)}
word_to_index = {k:i for i,k in enumerate(newVocab)}
```

```python
import re
import tensorflow as tf

PAD_ID, UNK_ID, BOS_ID, EOS_ID = 0, 1, 2, 3
# assume: newVocab[0]="<PAD>", [1]="<UNK>", [2]="<BOS>", [3]="<EOS>"
word_to_index = {w:i for i,w in enumerate(newVocab)}

def tokenize(text: str):
    return re.findall(r"\b\w+(?:'\w+)?\b|[.!?,-]", text.lower())

def to_ids(tokens):
    return [word_to_index.get(tok, UNK_ID) for tok in tokens]

def pad_or_trim(ids, L):
    ids = ids[:L]
    if len(ids) < L:
        ids = ids + [PAD_ID] * (L - len(ids))
    return ids

def preprocessing_layer(text):
    # 1) base ids (add EOS to mark sentence end before padding)
    base = to_ids(tokenize(text)) + [EOS_ID]

    # 2) pad/trim to fixed length
    src_ids = pad_or_trim(base, maxInDs)              # encoder input

    # 3) build decoder inputs/labels (teacher forcing)
    tgt_in  = pad_or_trim([BOS_ID] + base[:-1], maxInDs)  # <BOS> + tokens (shifted right)
    tgt_out = pad_or_trim(base, maxInDs)                  # next-token labels

    return {"src_ids": src_ids, "tgt_in": tgt_in, "tgt_out": tgt_out}
processed_ds_train = ds["train"].map(lambda x: preprocessing_layer(x["Text"]))
processed_ds_test  = ds["test"].map(lambda x: preprocessing_layer(x["Text"]))

tf_ds_train = processed_ds_train.to_tf_dataset(
    columns=["src_ids", "tgt_in"],
    label_cols=["tgt_out"],
    batch_size=10,
    shuffle=True,
)

tf_ds_test = processed_ds_test.to_tf_dataset(
    columns=["src_ids", "tgt_in"],
    label_cols=["tgt_out"],
    batch_size=10,
    shuffle=True,
)
```

```python
def shakesphereModel():
  inputs = tf.keras.layers.Input(shape = (maxInDs,))
  X = emb_layer(inputs)
  # LSTM Block 1
  X , _ , _ = LSTM(units = 256,activation = 'tanh' , return_sequences=True , return_state = True)(X)
  X = Dropout(0.4)(X)
  X , _ , _ = LSTM(units = 512,activation = 'tanh' , return_sequences=True ,return_state = True)(X)
  X = Dropout(0.4)(X)
  # X = LSTM(units = 256,activation = 'tanh' , return_sequences=True)(X)
  # X = Dropout(0.4)(X)
  X = Dense(units = vocab_size , activation = 'softmax')(X)
  outputs = X
  model = Model(inputs = inputs , outputs = outputs)
  return model
model = shakesphereModel()
```

```python
def shakespereKerasSelf():
    # Inputs for encoder and decoder
    encoder_inputs = tf.keras.layers.Input(name = "src_ids",shape = (maxInDs,))
    decoder_inputs = tf.keras.layers.Input(name = "tgt_in",shape = (maxInDs,))

    # Embedding layer for endoer and decoder inputs
    emb_layer = keras_hub.layers.TokenAndPositionEmbedding(
        vocabulary_size = vocab_size,
        sequence_length = maxInDs,
        embedding_dim = 300,
        embeddings_initializer="uniform",
        mask_zero=True,
    )

    # Encoder and decoder embeddings
    enc_emb = emb_layer(encoder_inputs)
    dec_emb = emb_layer(decoder_inputs)

    # Encoder Layer
    for _ in range(5):
        enc_emb = keras_hub.layers.TransformerEncoder(
            intermediate_dim = 1200,
            num_heads = 5,
            dropout=0.2,
            activation="relu",
            layer_norm_epsilon=1e-05,
            kernel_initializer="glorot_uniform",
            bias_initializer="zeros",
            normalize_first=False,
        )(enc_emb)
        
    # Encoder ouput
    encoder_outputs = enc_emb

    # Decoder Layer
    for _ in range(5):
        dec_emb = keras_hub.layers.TransformerDecoder(
            intermediate_dim = 1200,
            num_heads = 5,
            dropout=0.2,
            activation="relu",
            layer_norm_epsilon=1e-05,
            kernel_initializer="glorot_uniform",
            bias_initializer="zeros",
        )(dec_emb)

    # Decoder Output
    decoder_outputs = dec_emb

    outputs = tf.keras.layers.Dense(vocab_size)(decoder_outputs)

    return tf.keras.Model(inputs = [encoder_inputs,decoder_inputs] , outputs = outputs)

model = shakespereKerasSelf()
model.summary()
```

```python
import numpy as np
import tensorflow as tf

# Generated array init
generated_array = []

# States Init
a0 = c0 = np.zeros((1,256))
a1 = c1 = np.zeros((1,512))

# Pulling trained layers from the model
embedding_layer = model.layers[1]
lstm_layer_1 = model.layers[2]
lstm_layer_2 = model.layers[4]
dense_layer_1 = model.layers[6]

# Start tokens
inputs = tf.constant([[12]], dtype=tf.int32)  # start with first token

temperature = 0.7  # controls randomness (higher = more diverse)
top_k = 10         # optional: only consider top-k tokens

for i in range(40):
    # Embedding
    X = embedding_layer(inputs)

    # LSTM layers
    X, a0, c0 = lstm_layer_1(X, initial_state=[a0, c0])
    X, a1, c1 = lstm_layer_2(X, initial_state=[a1, c1])

    # Dense
    X = dense_layer_1(X)

    # Convert tensor to numpy
    logits = X.numpy()  # shape: (batch, seq_len, vocab_size)
    
    # Apply temperature and numerical stability
    scaled_logits = logits / temperature
    scaled_logits -= np.max(scaled_logits, axis=-1, keepdims=True)
    probs = np.exp(scaled_logits)
    probs /= probs.sum(axis=-1, keepdims=True)

    # Top-k sampling
    top_k_indices = np.argsort(probs[0, -1, :])[-top_k:]
    top_k_probs = probs[0, -1, :][top_k_indices]
    top_k_probs /= top_k_probs.sum()

    # Sample next token
    next_token = np.random.choice(top_k_indices, p=top_k_probs)

    # Append to generated array
    generated_array.append(next_token)

    # Prepare input for next step (last token only)
    inputs = tf.constant([[next_token]], dtype=tf.int32)

generated_array
```
