```python
def shakesphereModel():
  inputs = tf.keras.layers.Input(shape = (maxInDs,))
  X = emb_layer(inputs)
  # LSTM Block 1
  X , _ , _ = LSTM(units = 256,activation = 'tanh' , return_sequences=True , return_state = True)(X)
  X = Dropout(0.4)(X)
  X , _ , _ = LSTM(units = 512,activation = 'tanh' , return_sequences=True ,return_state = True)(X)
  X = Dropout(0.4)(X)
  # X = LSTM(units = 256,activation = 'tanh' , return_sequences=True)(X)
  # X = Dropout(0.4)(X)
  X = Dense(units = vocab_size , activation = 'softmax')(X)
  outputs = X
  model = Model(inputs = inputs , outputs = outputs)
  return model
model = shakesphereModel()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model

# ---- config ----
vocab_size = 32000         # set your vocab size
maxInDs    = 128           # sequence length (encoder + decoder)
d_model    = 300
num_heads  = 5             # must divide d_model
ffn_dim    = 2048
num_layers = 6
drop_p     = 0.1
PAD_ID     = 0

# ---- embeddings ----
def token_embedding_layer():
    # PAD_ID = 0 is masked
    return layers.Embedding(input_dim=vocab_size, output_dim=d_model, mask_zero=True)

def positional_embedding_layer():
    # trainable positional embeddings (0..maxInDs-1). Do NOT mask positions.
    return layers.Embedding(input_dim=maxInDs, output_dim=d_model)

emb_layer = token_embedding_layer()
pos_layer = positional_embedding_layer()

# ---- helper masks ----
def encoder_key_mask(src_ids):
    # src_ids: (b, s), True = keep, False = mask
    return tf.not_equal(src_ids, PAD_ID)[:, tf.newaxis, :]   # (b,1,s)

def look_ahead_mask(t):
    # (1, t, t) lower-triangular True = keep
    tri = tf.linalg.band_part(tf.ones((t, t), dtype=tf.bool), -1, 0)
    return tri[tf.newaxis, :, :]  # (1,t,t)

def decoder_self_mask(tgt_ids):
    # tgt_ids: (b, t)
    b = tf.shape(tgt_ids)[0]
    t = tf.shape(tgt_ids)[1]
    la = look_ahead_mask(maxInDs)[:, :t, :t]                 # (1,t,t)
    tgt_keep = tf.not_equal(tgt_ids, PAD_ID)[:, tf.newaxis, :]  # (b,1,t)
    # broadcast to (b,t,t): keep only past positions AND non-pad keys
    return tf.logical_and(la, tgt_keep)

def memory_mask_from_src(src_ids):
    # (b,1,s) → broadcasts to (b,t,s)
    return encoder_key_mask(src_ids)

# ---- encoder block ----
def encoder_block(X, enc_mask):
    attn = layers.MultiHeadAttention(
        num_heads=num_heads, key_dim=d_model // num_heads, dropout=drop_p,
        kernel_initializer='glorot_uniform', bias_initializer='zeros',
    )(X, X, attention_mask=enc_mask)
    attn = layers.Dropout(drop_p)(attn)
    X = layers.LayerNorm()(X + attn)

    Y = layers.Dense(ffn_dim)(X)
    Y = layers.Activation('relu')(Y)  # (option: 'gelu')
    Y = layers.Dropout(drop_p)(Y)
    Y = layers.Dense(d_model)(Y)
    Y = layers.Dropout(drop_p)(Y)
    X = layers.LayerNorm()(X + Y)
    return X

# ---- decoder block ----
def decoder_block(Y, enc_out, self_mask, mem_mask):
    # (1) masked self-attn
    attn1 = layers.MultiHeadAttention(
        num_heads=num_heads, key_dim=d_model // num_heads, dropout=drop_p,
        kernel_initializer='glorot_uniform', bias_initializer='zeros',
    )(Y, Y, attention_mask=self_mask)
    attn1 = layers.Dropout(drop_p)(attn1)
    Y = layers.LayerNorm()(Y + attn1)

    # (2) cross-attn
    attn2 = layers.MultiHeadAttention(
        num_heads=num_heads, key_dim=d_model // num_heads, dropout=drop_p,
        kernel_initializer='glorot_uniform', bias_initializer='zeros',
    )(Y, enc_out, attention_mask=mem_mask)
    attn2 = layers.Dropout(drop_p)(attn2)
    Y = layers.LayerNorm()(Y + attn2)

    # (3) FFN
    ff = layers.Dense(ffn_dim)(Y)
    ff = layers.Activation('relu')(ff)   # (option: 'gelu')
    ff = layers.Dropout(drop_p)(ff)
    ff = layers.Dense(d_model)(ff)
    ff = layers.Dropout(drop_p)(ff)
    Y = layers.LayerNorm()(Y + ff)
    return Y

# ---- full model ----
def shakesphereTransformer():
    # encoder inputs
    src_ids = layers.Input(shape=(maxInDs,), dtype=tf.int32)     # (b, s)
    # decoder inputs (shifted targets)
    tgt_ids = layers.Input(shape=(maxInDs,), dtype=tf.int32)     # (b, t)

    # positions (fixed [0..maxInDs-1])
    positions = tf.range(start=0, limit=maxInDs, delta=1)

    # encoder embeddings
    X = emb_layer(src_ids)                                       # (b, s, d)
    X = tf.cast(X, tf.float32) * tf.math.sqrt(tf.cast(d_model, tf.float32))
    pos_enc = pos_layer(positions)                               # (s, d)
    X = X + pos_enc                                              # broadcast to (b,s,d)

    # encoder mask
    enc_mask = encoder_key_mask(src_ids)                         # (b,1,s)

    # encoder stack
    for _ in range(num_layers):
        X = encoder_block(X, enc_mask)
    enc_out = X                                                  # (b, s, d)

    # decoder embeddings (share token emb; share pos layer)
    Y = emb_layer(tgt_ids)                                       # (b, t, d)
    Y = tf.cast(Y, tf.float32) * tf.math.sqrt(tf.cast(d_model, tf.float32))
    tgt_pos = pos_layer(positions)                               # (t, d)
    Y = Y + tgt_pos

    # decoder masks
    self_mask = decoder_self_mask(tgt_ids)                       # (b, t, t)
    mem_mask  = memory_mask_from_src(src_ids)                    # (b, 1, s)

    # decoder stack
    for _ in range(num_layers):
        Y = decoder_block(Y, enc_out, self_mask, mem_mask)

    # LM head
    logits = layers.Dense(vocab_size)(Y)                         # (b, t, vocab)

    return Model([src_ids, tgt_ids], logits)

# build
model = shakesphereTransformer()
model.summary()
```

```python
model.compile(loss = tf.keras.losses.SparseCategoricalCrossentropy(),optimizer = "adam" , metrics = [tf.keras.metrics.SparseCategoricalCrossentropy()])
model.fit(tf_ds_train,epochs=40)
```

```python
import numpy as np
import tensorflow as tf

# Generated array init
generated_array = []

# States Init
a0 = c0 = np.zeros((1,256))
a1 = c1 = np.zeros((1,512))

# Pulling trained layers from the model
embedding_layer = model.layers[1]
lstm_layer_1 = model.layers[2]
lstm_layer_2 = model.layers[4]
dense_layer_1 = model.layers[6]

# Start tokens
inputs = tf.constant([[12]], dtype=tf.int32)  # start with first token

temperature = 0.7  # controls randomness (higher = more diverse)
top_k = 10         # optional: only consider top-k tokens

for i in range(40):
    # Embedding
    X = embedding_layer(inputs)

    # LSTM layers
    X, a0, c0 = lstm_layer_1(X, initial_state=[a0, c0])
    X, a1, c1 = lstm_layer_2(X, initial_state=[a1, c1])

    # Dense
    X = dense_layer_1(X)

    # Convert tensor to numpy
    logits = X.numpy()  # shape: (batch, seq_len, vocab_size)
    
    # Apply temperature and numerical stability
    scaled_logits = logits / temperature
    scaled_logits -= np.max(scaled_logits, axis=-1, keepdims=True)
    probs = np.exp(scaled_logits)
    probs /= probs.sum(axis=-1, keepdims=True)

    # Top-k sampling
    top_k_indices = np.argsort(probs[0, -1, :])[-top_k:]
    top_k_probs = probs[0, -1, :][top_k_indices]
    top_k_probs /= top_k_probs.sum()

    # Sample next token
    next_token = np.random.choice(top_k_indices, p=top_k_probs)

    # Append to generated array
    generated_array.append(next_token)

    # Prepare input for next step (last token only)
    inputs = tf.constant([[next_token]], dtype=tf.int32)

generated_array
```
